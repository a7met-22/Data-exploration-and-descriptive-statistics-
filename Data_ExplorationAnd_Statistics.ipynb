## Frequency Distributions for Categorical Columns

### Subtask:
Analyze and visualize the frequency distribution of categorical columns ('Gender', 'Country', 'City', 'Signup_Quarter') using count plots or value counts to understand their composition and identify dominant categories.


**Reasoning**:
I will define the list of categorical columns and then iterate through them to print the value counts for each, which directly addresses the second instruction.



````python
categorical_cols = ['Gender', 'Country', 'City', 'Signup_Quarter']

print("Frequency Distributions for Categorical Columns:")
for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
````

**Reasoning**:
I will create count plots for each categorical column to visualize their frequency distributions, as specified in the instructions. This complements the value counts already displayed.



````python
import matplotlib.pyplot as plt
import seaborn as sns

categorical_cols = ['Gender', 'Country', 'City', 'Signup_Quarter']

plt.figure(figsize=(18, 12))
for i, col in enumerate(categorical_cols):
    plt.subplot(2, 2, i + 1) # Arrange plots in a 2x2 grid
    sns.countplot(data=df, x=col, palette='viridis')
    plt.title(f'Frequency Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability

plt.tight_layout()
plt.show()
````
